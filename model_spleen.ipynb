```python
path_train = "datas/spleen/mouse_spleen_1/mouse_spleen_1_Processed.h5ad"
path_val = "datas/spleen/mouse_spleen_2/mouse_spleen_2_Processed.h5ad"
if not (os.path.exists(path_train) and os.path.exists(path_val)):
    raise FileNotFoundError("Spleen processed h5ad files not found")
```

```python
ad_train = sc.read_h5ad(path_train, backed="r")
ad_val = sc.read_h5ad(path_val, backed="r")
```

```python
p_train = [str(x) for x in list(ad_train.uns.get("protein_names", []))]
p_val = [str(x) for x in list(ad_val.uns.get("protein_names", []))]
common_proteins = sorted(set(p_train).intersection(p_val)) or None
del ad_train, ad_val
```

```python
gc.collect()
```

```python
train_dataset = SpleenMultimodalDataset(path_train, target_proteins=common_proteins)
val_dataset = SpleenMultimodalDataset(path_val, target_proteins=common_proteins)
train_loader = DataLoader(train_dataset, batch_size=256, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=256, shuffle=False)
NUM_GENES = train_dataset.rna_data.shape[1]
NUM_PROTEINS = train_dataset.protein_data.shape[1]
print(f"Final Config: {NUM_GENES} genes, {NUM_PROTEINS} proteins")
```

```python
model_kwargs = {"num_genes": NUM_GENES}
models_to_train = MAIN_MODELS
```

```python
all_histories = {}
for i, (model_cls, name) in enumerate(models_to_train):
    print(f"\n[{i + 1}/{len(models_to_train)}] Training {name}")
    all_histories[name] = train_engine(model_cls, name, train_loader, val_loader, NUM_PROTEINS, device, epochs=30, model_kwargs=model_kwargs, lr=1e-4)
```
